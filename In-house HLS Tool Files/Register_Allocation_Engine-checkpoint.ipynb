```python
def application(file):
    minadder, minmulti = 1,1                                                         #r1min r2min  
    maxadder, maxmulti = 0,0                                                         #initialized with 0
    arAdder,arMulti, arRegister, delayAdder, delayMulti = 0,0,0,0,0                  #initialized with 0
    for i in file:
        if len(i)!=0:
            if i[0] == 'adder(16)' and maxadder<int(i[-1]):
                maxadder = int(i[-1])
                arAdder = eval(i[2][:-2])
                delayAdder = eval(i[3][:-2])
            elif i[0] == 'multiplier(16)' and maxmulti<int(i[-1]):
                maxmulti = int(i[-1])
                arMulti = eval(i[2][:-2])
                delayMulti = eval(i[3][:-2])
            elif i[0].find("register") != -1:
                arRegister = eval(i[0].split()[1][:-2])
    #print(arAdder,arMulti, arRegister, delayAdder, delayMulti)
    return (minadder, minmulti, maxadder, maxmulti, arAdder,arMulti, arRegister, delayAdder, delayMulti)
```

```python
class Node:
    def __init__(self, s):
        self.operator = s[0]
        self.operand_1 = int(s[1])
        self.operand_2 = int(s[2])
        self.id = int(s[3])
        self.control_step = None
```

```python
def read_file(s):
    file = open(s)
    fir = file.read()
    file.close()
    fir_dfg = ""
    for i in fir:
        if i != '\n':
            fir_dfg += i
    fir_dfg = fir_dfg.split(",")
    return fir_dfg
```

```python
def resource_constraint(sign, resources):
    list_of_nodes = []
    for i in range(int(len(sign)/4)):
        list_of_nodes.append(Node(sign[i*4:i*4+4]))
    operators = {}
    for i in list_of_nodes:
        operators[i.operator] = None
    for i in operators:
        if i == '+':
            operators[i] = resources[0]
        if i == '*':
            operators[i] = resources[1]
    return operators, list_of_nodes
```

```python
def register_allocation(list_of_nodes, max_con):
    graph_node = {}
    registers = {}
    node_c = {}
    counter = 0
    node_c[0] = 0
    for node in list_of_nodes:
        node_c[node.id] = node.control_step
    for i in list_of_nodes:
        if i.operand_1 == 0 and i.operand_2 == 0:
            graph_node[counter] = []
            registers[counter] = []
            graph_node[counter].append(0)
            registers[counter].append(0)
            registers[counter].append(0)
            counter += 1
    for i in list_of_nodes:
        if i.operand_1 == 0 and i.operand_2 == 0:
            graph_node[counter] = []
            registers[counter] = []
            graph_node[counter].append(0)
            registers[counter].append(0)
            registers[counter].append(0)
            counter += 1
    for i in list_of_nodes:
        if i.operand_1 == 0 and i.operand_2 != 0 or i.operand_1 != 0 and i.operand_2 == 0:
            graph_node[counter] = []
            registers[counter] = []
            graph_node[counter].append(0)
            registers[counter].append(i.operand_1)
            registers[counter].append(i.operand_2)
            counter += 1
            
    temp = counter - 1
    for i in list_of_nodes:
        graph_node[counter] = []
        registers[counter] = []
        graph_node[counter].append(i.control_step)
        if i.operand_1 != 0:
            registers[counter].append(i.operand_1+temp)
        else:
            registers[counter].append(i.id-1)
        if i.operand_2 != 0:
            registers[counter].append(i.operand_2+temp)
        else:
            registers[counter].append(i.id-1)
        for j in range(node_c[i.operand_1], i.control_step):
            if i.operand_1 != 0:    
                graph_node[i.operand_1+temp].append(j)
                graph_node[i.operand_1+temp] = list(set(graph_node[i.operand_1+temp]))
            else:
                graph_node[i.id-1].append(j)
                graph_node[i.id-1] = list(set(graph_node[i.id-1]))
        for j in range(node_c[i.operand_2], i.control_step):
            if i.operand_2 != 0:    
                graph_node[i.operand_2+temp].append(j)
                graph_node[i.operand_2+temp] = list(set(graph_node[i.operand_2+temp]))
            else:
                graph_node[i.id-1].append(j)
                graph_node[i.id-1] = list(set(graph_node[i.id-1]))
        counter += 1
    graph_node
    graph_node_c = {}
    for i in range(max_con+1):
        graph_node_c[i] = []
        for j in graph_node:
            if i in graph_node[j]:
                graph_node_c[i].append(j)
                
    graph_nodes = {}
    for i in range(len(graph_node)):
        graph_nodes[i] = []
        for j in range(len(graph_node)):
            if i!=j:
                if len(set(graph_node[j]).intersection(set(graph_node[i])))>0:
                    graph_nodes[i].append(j)
    edges = []
    for i in graph_nodes:
        for j in graph_nodes[i]:
            edges.append((i,j))
   
    return edges, graph_node_c, graph_nodes, temp, registers
```

```python
def register_coloring(graph_nodes):
    res = [-1 for x in range(len(graph_nodes))]
    available = [False for x in range(len(graph_nodes))]
    cn = 0
    for each_node in graph_nodes:
        for adj_node in graph_nodes[each_node]:
            if res[adj_node] != -1:
                available[res[adj_node]] = True
    
        cr = 0
        for no in range(len(available)):
            if available[no] == False:
                cr = no
                break
        res[each_node] = cr
        cn = max(cn,cr+1)
    
        for adj_node in graph_nodes[each_node]:
            if res[adj_node] != -1:
                available[res[adj_node]] = False
    return cn, res
```

```python
def register_recoloring(cn, register_c, graph_nodes, graph_node_c, temp, registers):
    register_c = [-1 for i in range(len(graph_nodes))]
    for cycle in graph_node_c:
        cc = 0
        cl = []
        if cycle == 0:
            for i in range(temp+1):
                register_c[i] = cc
                cc += 1
            continue
        for reg in graph_node_c[cycle]:
            if register_c[reg] == -1:
                if register_c[min(registers[reg])] not in cl:
                    register_c[reg] = register_c[min([register_c[each] for each in registers[reg] ])]
                    cl.append(register_c[min([register_c[each] for each in registers[reg] ])])
                else:
                    k = register_c[min(registers[reg])]
                    while k not in cl:
                        k += 1
                        if k == cn:
                            k /= cn
                    register_c[reg] = k
                    cl.append(k)
   
    return register_c
```

```python
def final_count(total_count, digital_template):
    no_of_zeros = 0
    no_of_ones = 0
    for i in range(total_count):
        if i< len(digital_template) and digital_template[i] == '1':
            no_of_ones += 1
        elif i < len(digital_template) and digital_template[i] == '0':
            no_of_zeros += 1
    #print(no_of_ones, no_of_zeros, point_count)
    ones = min(no_of_ones, max_1)
    zeros = min(no_of_zeros, max_0)
    
    return ones, zeros
```

```python
def AETP(position, swarm_size, dim, L, U):
    for i in range(swarm_size):
        for j in range(dim):
            while(position[i][j] < L[j]):
                position[i][j] = L[j]
            while (position[i][j] > U[j]):
                position[i][j] = U[j]
    return position
```

```python
def PVC(velocity, swarm_size, dim, n_vmax, p_vmax):
    for i in range(swarm_size):
        for j in range(dim):
            if n_vmax > velocity[i][j]:
                velocity[i][j] = n_vmax
            if p_vmax < velocity[i][j]:
                velocity[i][j] = p_vmax
    return velocity
```

```python
def ARM(position, swarm_size, dim):
    for i in range(swarm_size):
        if (i%2==1): # Left Rotation 
            for j in range(dim-1):
                Temp = position[i][j] 
                position[i][j]= position[i][j+1]      #j+1 i.e. dimension-1
                position[i][j+1]= Temp
                
        if(i%2==0):
            X = random.randint(-3,3)
            for j in range(dim):
                position[i][j] = position[i][j] + X #X is a random number between [1,3]   

    return position
```
